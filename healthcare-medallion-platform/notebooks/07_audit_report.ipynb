```python
from pyspark.sql import functions as F

CATALOG = "healthcare_medallion_dbw"

audit = spark.table(
    f"{CATALOG}.ops.pipeline_audit_log"
)

display(audit)
```

```python
latest_runs = (
    audit
    .select(
        "batch_id",
        "source_name",
        "layer",
        "pipeline_start_time",
        "pipeline_end_time",
        "rows_read",
        "rows_written",
        "rows_rejected",
        "status",
        "pipeline_duration_secs",
        "environment",
        "sla_met"
    )
    .orderBy(
        F.col("pipeline_start_time").desc()
    )
)

display(latest_runs)
```

```python
status_summary = (
    audit
    .groupBy("status")
    .agg(
        F.count("*").alias("run_count")
    )
)

display(status_summary)
```

```python
rows_by_source = (
    audit
    .groupBy(
        "source_name"
    )
    .agg(
        F.sum("rows_read").alias(
            "total_rows_read"
        ),
        F.sum("rows_written").alias(
            "total_rows_written"
        ),
        F.sum("rows_rejected").alias(
            "total_rows_rejected"
        )
    )
    .orderBy(
        F.col("total_rows_read").desc()
    )
)

display(rows_by_source)
```

```python
duration_summary = (
    audit
    .groupBy(
        "source_name"
    )
    .agg(
        F.round(
            F.avg(
                "pipeline_duration_secs"
            ),
            2
        ).alias(
            "avg_duration_seconds"
        )
    )
)

display(duration_summary)
```

```python
sla_summary = (
    audit
    .groupBy(
        "sla_met"
    )
    .agg(
        F.count("*").alias("run_count")
    )
)

display(sla_summary)
```

```python
(
    rows_by_source.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable(
        f"{CATALOG}.gold.audit_summary"
    )
)
```
